```python
from unet import UNetAutoencoder
from classifier import Classifier
from dataset import SemiSupervisedMnistDataModule
from train import train

from hydra import initialize, compose
from omegaconf import OmegaConf
import random, numpy as np
```

```python
# Ajusta config_path si tu carpeta conf está en otra ruta
with initialize(config_path="../config", version_base=None):
    cfg = compose(
        config_name="config",
        # aquí puedes sobrescribir parámetros si quieres
        overrides=[
            # e.g.: "batch_size=32", "optimizer.name=SGD", "learning_rate=0.01"
        ]
    )

# Opcional: imprime la config para verificar
print(OmegaConf.to_yaml(cfg))

# Fija semillas
random.seed(cfg.seed)
np.random.seed(cfg.seed)
```

```python
# Configuraciones básicas
MAX_EPOCHS          = cfg.max_epochs
EARLY_STOP_PATIENCE = cfg.early_stop_patience
PROJECT_NAME        = cfg.project_name
```

## Experimento 70% sin labels y 30% con labels

```python
dm_70 = SemiSupervisedMnistDataModule(
    data_dir="../data/",
    batch_size=cfg.batch_size,
    num_workers=cfg.num_workers,
    label_pct=0.3 
)
dm_70.setup()

dm_90 = SemiSupervisedMnistDataModule(
    data_dir="../data/",
    batch_size=cfg.batch_size,
    num_workers=cfg.num_workers,
    label_pct=0.1
)
dm_90.setup()
```

### UNetAutoencoders

```python
autoencoder_70 = UNetAutoencoder()

train(autoencoder_70, dm_70.unlabeled_dataloader(), dm_70, max_epochs=MAX_EPOCHS, early_stop_patience=EARLY_STOP_PATIENCE, project_name=PROJECT_NAME, run_name="autoencoder_70", checkpoint_dir="../output/checkpoints/autoencoder/70")
```

```python
autoencoder_90 = UNetAutoencoder()

train(autoencoder_90, dm_90.unlabeled_dataloader(), dm_90, max_epochs=MAX_EPOCHS, early_stop_patience=EARLY_STOP_PATIENCE, project_name=PROJECT_NAME, run_name="autoencoder_90", checkpoint_dir="../output/checkpoints/autoencoder/90")
```

### Cargar Checkpoint (opcional) 

```python
pretrained_autoencoder_70 = UNetAutoencoder.load_from_checkpoint("../output/checkpoints/unet/70/epoch=18-val_loss=0.00022231.ckpt")
pretrained_autoencoder_90 = UNetAutoencoder.load_from_checkpoint("../output/checkpoints/unet/90/epoch=18-val_loss=0.00016913.ckpt")
raw_autoencoder = UNetAutoencoder()
```

### Clasificador con encoder UNet original

```python
classifier_a_70 = Classifier(encoder=raw_autoencoder.get_encoder(), learning_rate=cfg.learning_rate, freeze_encoder=False)

train(classifier_a_70, dm_70.labeled_dataloader(), dm_70, max_epochs=MAX_EPOCHS, early_stop_patience=EARLY_STOP_PATIENCE, project_name=PROJECT_NAME, run_name="classifier_a_70", checkpoint_dir="../output/checkpoints/classifierA/70")
```

```python
classifier_a_90 = Classifier(encoder=raw_autoencoder.get_encoder(), learning_rate=cfg.learning_rate)

train(classifier_a_90, dm_90.labeled_dataloader(), dm_90, max_epochs=MAX_EPOCHS, early_stop_patience=EARLY_STOP_PATIENCE, project_name=PROJECT_NAME, run_name="classifier_a_90", checkpoint_dir="../output/checkpoints/classifierA/90")
```

### Clasificador con encoder UNet preentrenado y congelado

```python
classifier_b1_70 = Classifier(encoder=pretrained_autoencoder_70.get_encoder(), learning_rate=cfg.learning_rate, freeze_encoder=True)

train(classifier_b1_70, dm_70.labeled_dataloader(), dm_70, max_epochs=MAX_EPOCHS, early_stop_patience=EARLY_STOP_PATIENCE, project_name=PROJECT_NAME, run_name="classifier_b1_70", checkpoint_dir="../output/checkpoints/classifierB1/70")
```

```python
classifier_b1_90 = Classifier(encoder=pretrained_autoencoder_90.get_encoder(), learning_rate=cfg.learning_rate, freeze_encoder=True)

train(classifier_b1_90, dm_90.labeled_dataloader(), dm_90, max_epochs=MAX_EPOCHS, early_stop_patience=EARLY_STOP_PATIENCE, project_name=PROJECT_NAME, run_name="classifier_b1_90", checkpoint_dir="../output/checkpoints/classifierB1/90")
```
